```python
from lfs.data import get_patches, load_mnist
from lfs.selection import loco
from lfs.selection import minshap
from lfs.viz import plot_patch_importance, plot_selected
from lfs.models import logistic_regression
from lfs.models import neural_net

from sklearn.metrics import accuracy_score
```

```python
X_train, X_test, y_train, y_test = load_mnist()
```

```python
clf = logistic_regression(X_train, y_train)
```

```python
nn = neural_net(X_train, y_train)
```

```python
patches = get_patches(patch_size=7)
scores = loco(clf, patches, X_train, X_test, y_train, y_test)
plot_patch_importance(scores, patches)
```

```python
patches = get_patches(patch_size=7)
scores = loco(nn, patches, X_train, X_test, y_train, y_test, n_jobs=1)
plot_patch_importance(scores, patches)
```

```python
patches = get_patches(patch_size=7)  # 16 patches
minshap_result = minshap(nn, patches, X_train, y_train, 
                         loss='zero_one', K=10, n_jobs=1)
plot_selected(minshap_result, patches)
```

```python
patches = get_patches(patch_size=4)  # 16 patches
minshap_result = minshap(nn, patches, X_train, y_train, 
                         loss='zero_one', K=10, n_jobs=1)
fig, axes = plot_selected(minshap_result, patches)
fig.savefig("minshap_selected_patches_4.png", dpi=300)
```
